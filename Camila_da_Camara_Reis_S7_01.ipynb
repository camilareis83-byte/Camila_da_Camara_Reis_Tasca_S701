## ✅ Resultado:

*1. Pedir los datos al usuario vía input. `obtener_datos()`*

---

Consiste en:
1. Nombrar las variables que utilizaremos en el ejercicio.
2. Escribir frases que permitan al usuario comprender qué tiene que contestar.
3. Devolver y guardar las variables.

````python
def obtener_datos():
    nombre_apellido = input("Introduzca su nombre y apellido: ")
    telefono = input("Introduzca su teléfono de contacto (nueve dígitos): ")
    anos_experiencia = input("Indique cuántos años de experiencia tienes en la empresa: ")
    nivel_catalan = input("Indique su Conocimiento de Catalán (0-3 puntos):\n0 = sin conocimiento\n1 = hasta nivel B1\n2 = nivel B2\n3 = C1 o superior\n")
    nivel_excel = input("Indique su Dominio de Excel (0-3 puntos):\n0 = nivel básico\n1 = uso de funciones\n2 = tablas dinámicas\n3 = Visual Basic y Macros\n")
    nivel_academico = input("Indique qué título académico posee (0-3 puntos):\n0 = ESO\n1 = FP\n2 = Universidad\n3 = Máster o superior\n")
    return nombre_apellido, telefono, anos_experiencia, nivel_catalan, nivel_excel, nivel_academico
````

*2. Convertir estos datos a números enteros: Años de experiencia dentro de la empresa, Conocimiento de Catalán, Dominio de Excel y Título académico.  `convertir_datos(anos_experiencia, nivel_catalan, nivel_excel, nivel_academico)`*

---

1. Consiste en:
    - Convertir los datos de anos_experiencia, nivel_catalan, nivel_excel, nivel_academico a tipo *integer*.
    - Devolver los valores.
2. Si alguno de los pasos falla, se muestra un mensaje de error.


````python
def convertir_datos(anos_experiencia, nivel_catalan, nivel_excel, nivel_academico):
    try:
        anos_experiencia = int(anos_experiencia)
        nivel_catalan = int(nivel_catalan)
        nivel_excel = int(nivel_excel)
        nivel_academico = int(nivel_academico)
        return anos_experiencia, nivel_catalan, nivel_excel, nivel_academico
    except ValueError:
        print("Error: Por favor, introduzca valores enteros válidos para Años de Experiencia, Conocimiento de Catalán, Dominio de Excel y Título Académico")
        return None, None, None, None
````

*3. Asegurar de que toda la información introducida es correcta y entra dentro de los valores permitidos: `validar_datos()`*
  * *Nombre y Apellidos: garantizar que no sea una entrada de datos vacía*
  * *Teléfono de contacto: garantizar que tenga 9 dígitos*
  * *Años de experiencia dentro de la empresa: garantizar que sea un entero entre 0 y 60*
  * *Conocimiento de Catalán: garantizar que sea un entero entre 0 y 3*
  * *Dominio de Excel: garantizar que sea un entero entre 0 y 3*
  * *Título académico: garantizar que sea un entero entre 0 y 3*

---

Consiste en:

1. Si `nombre_apellido` es un string vacío, muestra un mensaje de error. El `.strip()`elimina el problema de strings solamente con espacios.

2. Si `telefono` no tiene exactamente 9 dígitos numéricos (como son los teléfonos en España), muestra un mensaje de error.

3. Si `anos_experiencia` no está entre 0 y 60, muestra un mensaje de error. Aquí se tiene en cuenta que es poco probable que haya empleados con más de 60 años de experiencia en la empresa, y que personas con 0 años de experiencia pueden optar al puesto, pero el cálculo de la puntuación total se verá afectado.

4. Si `nivel_catalan`, `nivel_excel` y `nivel_academico` no son números entre 0 y 3, muestra un mensaje de error.


````python
def validar_datos(nombre_apellido, telefono, anos_experiencia, nivel_catalan, nivel_excel, nivel_academico):
    nombre_valido = nombre_apellido.strip() != ""
    telefono_valido = telefono.isdigit() and len(telefono) == 9
    experiencia_valida = 0 <= anos_experiencia <= 60
    niveles_validos = all(0 <= nivel <= 3 for nivel in (nivel_catalan, nivel_excel, nivel_academico))

    if nombre_valido and telefono_valido and experiencia_valida and niveles_validos:
        return True

    else:
        print("Error: Por favor, introduzca datos válidos.")
        return False
````

*4. Calcular una puntuación total aplicando el peso de la experiencia (años * 1.5) y sumando los puntos de catalán, excel y titulación. `calcular_puntuacion()`*

---

Consiste en:
1. Calcula la suma total de puntuación (anos_experiencia * 1.5 + nivel_catalan + nivel_excel + nivel_academico) y almacena en resultado_pont.
2. Retorna resultado_pont.

````python
def calcular_puntuacion(anos_experiencia, nivel_catalan, nivel_excel, nivel_academico):
    resultado_pont = (anos_experiencia * 1.5) + nivel_catalan + nivel_excel + nivel_academico
    return resultado_pont
````

*5. Clasificar automáticamente a cada persona según los umbrales establecidos por la empresa: `clasificar_candidato()`*

* *Prioridad Alta → puntuación ≥ 12*
* *Candidato adecuado → 8 ≤ puntuación < 12*
* *No prioritario → puntuación < 8*

---

Consiste en:

1. Clasificar la variable resultado_pont según los rangos establecidos:
  * Si resultado_pont es menor que 8, devuelve "No prioritario".
  * Si resultado_pont está entre 8 (inclusive) y 12 (exclusivo), devuelve "Candidato Adecuado".
  * Si no se cumple ninguna de las condiciones anteriores (es decir, si la puntuación es 12 o más), devuelve "Prioridad Alta".


````python
def clasificar_candidato(resultado_pont):
    if resultado_pont < 8:
        return "No prioritario"
    elif 8 <= resultado_pont < 12:
        return "Candidato Adecuado"
    return "Prioridad Alta"
````

*6. Mostrar la puntuación individual al usuario. `mostrar_individual()`*

---

Consiste en:
1. Imprimir en pantalla los datos nombre_apellido, telefono y resultado_pont para el usuario.


````python
def mostrar_individual(nombre_apellido, telefono, resultado_pont):
    print(f"Nombre y Apellidos: {nombre_apellido}\nTeléfono de contacto: {telefono}\nTu resultado: {resultado_pont}")
````

7. *Cuando ya se hayan inscrito exactamente diez personas, la herramienta debe dejar de aceptar nuevos registros.*

---

Consiste en incluir etapas en la función `main_seleccion`():
1. Crear una lista vacía `candidatos` donde se añadirán los datos de cada candidato después de su validación.
2. Añadir los datos validados de cada candidato a la lista `candidatos` utilizando `.append` y guardarlos como un diccionario dentro de la lista, para generar después el informe final.
3. Crear un bucle while que contará los datos validados añadidos a la lista `candidatos` hasta llegar a un total de 10.

*8. Generar un informe final en pantalla, ordenado primero según la prioridad y, en caso de empate, por la puntuación obtenida. Este informe debe incluir el nombre, el teléfono y la puntuación de cada candidato. `mostrar_informe()`*

---

Consiste en:

1. Definir un diccionario de prioridad (ORDEN_PRIORIDAD), que asigna un valor numérico a cada nivel de prioridad ("Prioridad Alta": 0, "Candidato Adecuado": 1, "No prioritario": 2).

2. Ordenar y almacenar los datos de la lista `candidatos` en la variable `ordenados` sin modificar la lista original.

3. Utilizar el `sorted()` para ordenar `candidatos`:
  * key=lambda co: Define un criterio de ordenación a medida analizando cada candidato (co) de uno en uno.
  * Primer criterio (ORDEN_PRIORIDAD[...]): Ordena a los candidatos de mayor a menor prioridad (primero los de Prioridad Alta con valor 0, luego Candidato Adecuado con 1, y al final No prioritario con 2).
  * Segundo criterio (-co["puntuacion"]): En caso de empate en la prioridad, el signo menos (-) ordena las puntuaciones de forma descendente (de mayor a menor) para decidir quién va primero.

4. Imprimir el informe final de los candidatos ordenados: recorrer la lista ya ordenada (for candidato in ordenados) y mostrar en pantalla un informe con el nombre, teléfono, puntuación y clasificación de cada candidato separados por barras verticales (|).


````python
ORDEN_PRIORIDAD = {"Prioridad Alta": 0, "Candidato Adecuado": 1, "No prioritario": 2}

def mostrar_informe(candidatos):
    ordenados = sorted(candidatos, key=lambda co: (ORDEN_PRIORIDAD[co["clasificacion"]], -co["puntuacion"]))

    print("INFORME FINAL")
    for candidato in ordenados:
        print(f"{candidato['nombre']} | {candidato['telefono']} | {candidato['puntuacion']} | {candidato['clasificacion']}")
````

Construir la main_seleccion(). Esta es la función principal, que actúa como el orquestador de todo el proceso de inscripción y selección.

▶ `main_seleccion()`:

1. `candidatos`: se crea una lista vacía para almacenar los candidatos que cumplan con todos los requisitos de validación.
2. `while len(candidatos) < 10`: se inicia el bucle que se repetirá constantemente hasta que hayamos guardado con éxito exactamente 10 candidatos válidos en la lista.
3. `obtener_datos()`: almacena las variables que nos informa el candidato via inputs.
4. `convertir_datos(anos_experiencia, nivel_catalan, nivel_excel, nivel_academico)`: convierte los valores numéricos introducidos por el usuario de texto (string) a números enteros (int).
5. Si `anos_experiencia`, `nivel_catalan`, `nivel_excel` y `nivel_academico` son números enteros (Verifica que la conversión numérica haya sido exitosa. Si hubo un error en la conversión, ignora el resto de la iteración y vuelve a empezar), entonces:
  6. Si `validar_datos`(nombre_apellido, telefono, anos_experiencia, nivel_catalan, nivel_excel, nivel_academico) devuelve `True` (si los datos son válidos y están dentro de los valores permitidos), entonces:
      7. `calcular_puntuacion(anos_experiencia, nivel_catalan, nivel_excel, nivel_academico)`: calcula la nota final del candidato basándose en los pesos asignados y la almacena en `resultado_pont`.
      8. `clasificar_candidato(resultado_pont)`: clasifica el perfil según la puntuación obtenida ("Prioridad Alta", "Candidato Adecuado" o "No prioritario") y lo almacena en `clasificacion`.
      9. `mostrar_individual(nombre_apellido, telefono, resultado_pont)`: imprime en pantalla de forma inmediata los resultados individuales del candidato que se acaba de registrar.
      10. `candidatos.append()`: guarda la información de este candidato estructurada como un diccionario dentro de nuestra lista.

      11. `len(candidatos) = 10`: Una vez se alcanzan los 10 candidatos registrados, el bucle while termina y el programa imprime el mensaje "Número máximo de candidatos alcanzado.".
      12. `mostrar_informe(candidatos)`: ordenar a todos los candidatos y muestra el reporte final formateado.




````python
def main_seleccion():

    candidatos = []

    while len(candidatos) < 10:

        nombre_apellido, telefono, anos_experiencia, nivel_catalan, nivel_excel, nivel_academico = obtener_datos()
        anos_experiencia, nivel_catalan, nivel_excel, nivel_academico = convertir_datos(anos_experiencia, nivel_catalan, nivel_excel, nivel_academico)

        if anos_experiencia is not None:
            if validar_datos(nombre_apellido, telefono, anos_experiencia, nivel_catalan, nivel_excel, nivel_academico):
                resultado_pont = calcular_puntuacion(anos_experiencia, nivel_catalan, nivel_excel, nivel_academico)
                clasificacion = clasificar_candidato(resultado_pont)
                mostrar_individual(nombre_apellido, telefono, resultado_pont)

                candidatos.append({
                    "nombre": nombre_apellido,
                    "telefono": telefono,
                    "puntuacion": resultado_pont,
                    "clasificacion": clasificacion,
                })

    print("Número máximo de candidatos alcanzado.")

    mostrar_informe(candidatos)
````

2. `obtener_factor_edad()`:

---

Consiste en:

1. Crear un diccionario (`FACTORES_EDAD`) utilizando «Factors segons grup d'edat» creado por el CRM:
- Clave: "grupo_edad"
- Valor: "factores"

- "41-60": 0.85

````python
FACTORES_EDAD = {"18-25": 1.20, "26-40": 1.00, "41-60": 0.85, "60+": 0.70}

def obtener_factor_edad(grupo_edad):
    return FACTORES_EDAD[grupo_edad]
````

3. `calcular_riesgo()`: Utilizar factor_codigo y factor_edad, definidos anteriormente, para obtener el riesgo_final.

---

Consiste en:
1. Recuperar los valores de factor_codigo y factor_edad definidos anteriormente para realizar el cálculo necesario.
2. Aplicar la fórmula proporcionada por el CRM:
   - riesgo_final = factor_codigo * factor_edad
   - Redondear el resultado final a un máximo de 4 decimales.
3. Devolver el resultado del cálculo para almacenarlo.

````python
def calcular_riesgo(factor_codigo, factor_edad):
    return round(factor_codigo * factor_edad, 4)
````

4. `procesar_clientes()`: utilizar las tres funciones anteriores para procesar los datos disponibles en [clientes].

---

Consiste en:

1. Crear una lista vacía llamada `clientes_procesados`.
2. Procesar el conjunto de datos de los clientes:
- Recuperar los valores de id_cliente, nombre, edad, grupo_edad y codigo_comportamiento de la **lista de datos original**.
- Obtener segmento, mensaje y factor_codigo a través de `obtener_segmento_y_factor()`.
- Obtener factor_edad a través de `obtener_factor_edad()`.
- Obtener riesgo_final a través de `calcular_riesgo()`.
3. Guardar los datos procesados en la lista `clientes_procesados`.

````python
def procesar_clientes(clientes):

    clientes_procesados = []

    for cliente in clientes:
        segmento, mensaje, factor_codigo = obtener_segmento_y_factor(cliente["codigo_comportamiento"])
        factor_edad = obtener_factor_edad(cliente["grupo_edad"])

        clientes_procesados.append({
            "id_cliente": cliente["id_cliente"],
            "nombre": cliente["nombre"],
            "edad": cliente["edad"],
            "grupo_edad": cliente["grupo_edad"],
            "codigo_comportamiento": cliente["codigo_comportamiento"],
            "segmento": segmento,
            "mensaje": mensaje,
            "factor_codigo": factor_codigo,
            "factor_edad": factor_edad,
            "riesgo_final": calcular_riesgo(factor_codigo, factor_edad),
        })

    return clientes_procesados
````

  * Esta elección se debe a la ausencia del valor de compra en los datos de los clientes; por tanto, fue necesario seleccionar otro factor para definir al cliente más rentable de recuperar. Se eligió aquel con mayor riesgo de abandono.

````python
UMBRAL_RIESGO_BAJO = 0.25

def mostrar_respuestas(clientes_procesados):
    riesgos = []
    for cliente in clientes_procesados:
        riesgos.append((cliente["riesgo_final"], cliente["nombre"], cliente["segmento"]))

    print("1. Orden de contacto en la campaña de retargeting:")
    for riesgo, nombre, segmento in sorted(riesgos, reverse=True):
        print(f"   {nombre} - {segmento} - riesgo {riesgo}")

    print("\n2. Mismo comportamiento, distinta edad:")
    for codigo in range(5):
        print(f"   Código {codigo}:")
        for cliente in clientes_procesados:
            if cliente["codigo_comportamiento"] == codigo:
                print(f"      {cliente['nombre']} ({cliente['grupo_edad']}): {cliente['riesgo_final']}")

    sin_interes = []
    for riesgo, nombre, segmento in riesgos:
        if riesgo < UMBRAL_RIESGO_BAJO:
            sin_interes.append(nombre)
    print(f"\n3. Sin interés (riesgo < {UMBRAL_RIESGO_BAJO}): {sin_interes}")

    riesgo, nombre, segmento = max(riesgos)
    print(f"\n4. Más rentable de recuperar: {nombre} ({segmento}, riesgo {riesgo})")
````
